# Data Understanding

Load the raw CSVs, join them into a single analysis table, save the joined view to `data/interim/`, and summarize the dataset.

```python
from pathlib import Path

import pandas as pd

raw_dir = Path("../data/raw")
interim_dir = Path("../data/interim")
interim_dir.mkdir(parents=True, exist_ok=True)

impressions = pd.read_csv(raw_dir / "impressions.csv")
user_profile = pd.read_csv(raw_dir / "user_profile.csv")
product_catalog = pd.read_csv(raw_dir / "product_catalog.csv")

joined = (
    impressions
    .merge(user_profile, on="user_id", how="left")
    .merge(product_catalog, on="product_id", how="left")
)

joined_path = interim_dir / "joined_dataset.csv"
joined.to_csv(joined_path, index=False)

print(f"Saved joined dataset to: {joined_path}")
print(f"Shape: {joined.shape}")
print("\nColumn types:")
print(joined.dtypes)
print("\nMissing values:")
print(joined.isna().sum())
print(f"\nClick rate: {joined['clicked'].mean():.4f}")

joined.head()

```

## Notes

- The join keeps all impressions and enriches them with user and product attributes.
- The saved view lives at `data/interim/joined_dataset.csv`.
