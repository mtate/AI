# Baselines

This notebook computes simple benchmark rules on the test set so we have a baseline for accuracy and F1 score.

## Planned Steps

- Load the saved test set
- Evaluate a majority-class baseline
- Evaluate a simple CreditScore threshold rule
- Report accuracy and F1 score for both rules

```python
from pathlib import Path

import pandas as pd
from sklearn.metrics import accuracy_score, f1_score

TEST_PATH = Path('data/processed/encoded_test.csv')
if not TEST_PATH.exists():
    TEST_PATH = Path('../data/processed/encoded_test.csv')

test_df = pd.read_csv(TEST_PATH)

y_true = test_df['Approval']
always_yes_pred = pd.Series(['Yes'] * len(test_df), index=test_df.index)
threshold_pred = pd.Series(
    ['Yes' if score >= 680 else 'No' for score in test_df['CreditScore']],
    index=test_df.index,
)

results = pd.DataFrame([
    {
        'baseline': 'Always Yes',
        'accuracy': accuracy_score(y_true, always_yes_pred),
        'f1_score': f1_score(y_true, always_yes_pred, pos_label='Yes'),
    },
    {
        'baseline': 'CreditScore >= 680',
        'accuracy': accuracy_score(y_true, threshold_pred),
        'f1_score': f1_score(y_true, threshold_pred, pos_label='Yes'),
    },
])

print(results.to_string(index=False))
```
