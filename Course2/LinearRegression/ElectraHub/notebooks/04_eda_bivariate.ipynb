## Advertising Expenditure vs Sales

Raw scatter plot to inspect the overall relationship shape between:
- `Advertising_Expenditure` (x-axis)
- `Sales` (y-axis)


```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
```

```python
clean_path = Path('../data/processed/ElectraHub_Data_cleaned.csv')
raw_path = Path('../data/raw/ElectraHub_Data.csv')

data_path = clean_path if clean_path.exists() else raw_path
df = pd.read_csv(data_path)

print(f'Loaded: {data_path.name}')
print(f'Rows: {len(df):,} | Columns: {df.shape[1]}')
```

```python
plt.figure(figsize=(9, 6))
plt.scatter(
    df['Advertising_Expenditure'],
    df['Sales'],
    alpha=0.7,
    s=30,
    color='steelblue'
)
plt.title('Advertising Expenditure vs Sales (Raw Scatter)')
plt.xlabel('Advertising_Expenditure')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

## Sales by Popularity (Box Plot)

Box plot to compare `Sales` distributions across `Popularity` levels in business order:
`Very Low` → `Low` → `Moderate` → `High` → `Very High`.


```python
popularity_order = ['Very Low', 'Low', 'Moderate', 'High', 'Very High']

plot_df = df.copy()
plot_df['Popularity'] = plot_df['Popularity'].astype(str).str.strip()

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=plot_df,
    x='Popularity',
    y='Sales',
    order=popularity_order,
    hue='Popularity',
    palette='Set2'
)
plt.title('Sales Distribution by Popularity Level')
plt.xlabel('Popularity')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

## Sales by Campaign Type (Box Plot)

Box plot to compare `Sales` distributions across different `Campaign_Type` categories.


```python
campaign_order = ['Email', 'Search Ad', 'Social Media', 'Influencer']

plot_df = df.copy()
plot_df['Campaign_Type'] = plot_df['Campaign_Type'].astype(str).str.strip()

# Keep only campaign types present in data, preserving preferred order.
order_in_data = [c for c in campaign_order if c in set(plot_df['Campaign_Type'])]

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=plot_df,
    x='Campaign_Type',
    y='Sales',
    order=order_in_data,
    hue='Campaign_Type',
    palette='Set3'
)
plt.title('Sales Distribution by Campaign Type')
plt.xlabel('Campaign_Type')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```
